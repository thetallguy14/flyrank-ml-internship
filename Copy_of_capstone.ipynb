## 1. Question

*The research question and the decision it supports.*

```python
## Research Question

Which published content pages should be prioritized for content refresh based on
search performance and content quality signals?

This project supports SEO decision-making by ranking pages according to their refresh opportunity score.
Instead of manually reviewing thousands of pages, the model identifies high-impact pages using historical
search performance and content characteristics.
```

## 2. Data

*Which release, which tables, date windows, what you excluded and why. Public-safe.*

```python
## Data

This project uses the FlyRank ML Internship Warehouse dataset (release v20260703).

Tables used:

- fact_content_daily_performance_sample
- dim_content

The daily performance table provides Google Search Console and Google Analytics metrics including impressions, clicks, engagement and positions.

The content dimension provides metadata such as search volume, backlinks, competition, content length and publication status.

Only public-safe hashed identifiers were used.

No client names, URLs or private search queries were included.

The analysis uses approximately 50,000 sampled rows for efficient experimentation.
```

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

```python
## Methodology

The objective was to estimate which content pages offer the highest opportunity for content refresh.

Feature engineering combined search performance with content metadata.

Important features include:

- Average Position
- Search Volume
- Competition
- Backlinks
- Word Count
- Character Count
- Click Through Rate (CTR)
- Engagement Rate

The target variable was Opportunity Score:

Opportunity Score = Google Impressions − Google Clicks

Random Forest Regressor was selected because it models nonlinear relationships and automatically captures feature interactions.

The dataset was divided into 80% training and 20% testing.

Performance was evaluated using MAE, RMSE and R².
```

## 4. Results (vs baseline)

*Model vs baseline on the same split. The honest table.*

```python
## Results

Model Performance

MAE: 7.06

RMSE: 76.59

R² Score: 0.2613

The model successfully ranked high-opportunity pages and identified important predictors.

Feature importance analysis showed CTR as the strongest predictor followed by average position, character count and word count.

The recommendation engine generated ranked content refresh opportunities for SEO decision support.
```

## 5. Limitations

*What this work cannot claim.*

```python
## Limitations

This project identifies statistical relationships rather than causal effects.

The Opportunity Score is based on historical observations and should support—not replace—editorial judgment.

The sample dataset may not fully represent the complete warehouse.

No conclusions are made about Google's ranking algorithm.
```

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*

```python
## Ranked Recommendations

Priority actions include:

1. Refresh pages with high impressions but low CTR.
2. Improve metadata for highly visible pages.
3. Expand short content with low engagement.
4. Review pages with poor average ranking positions.
5. Monitor high-performing pages to preserve performance.

The recommendation engine automatically ranks pages by predicted opportunity score to support content strategy.
```
